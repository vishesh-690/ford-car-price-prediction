```python
import streamlit as st
import pandas as pd

# Load dataset
df = pd.read_csv("ford.csv")

# Sidebar filters
selected_model = st.sidebar.selectbox(
    "Select Car Model",
    df['model'].unique(),
    key="car_model_select"
)

selected_fuel = st.sidebar.selectbox(
    "Select Fuel Type",
    df['fuelType'].unique(),
    key="fuel_type_select"
)

# Apply filters
filtered_df = df[
    (df['model'] == selected_model) &
    (df['fuelType'] == selected_fuel)
]

st.write("📊 Filtered Data", filtered_df.head())
```
